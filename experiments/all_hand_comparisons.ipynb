# Compare all possible private hands

This notebook creates:

- a 1,326-entry vector comparing one private hand with every possible private hand;
- a 1,326 × 1,326 matrix comparing every possible private hand with every other hand.

Results are `0` (loss), `1` (tie), `2` (win), or `I` (incompatible). Matrix results are from the row hand's perspective.

```python
import sys
from pathlib import Path

project_root = next(
    folder for folder in (Path.cwd(), *Path.cwd().parents)
    if (folder / "cards.py").is_file()
)
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from cards import hand_names, make_hand
from showdown import PRIVATE_HANDS, comparison_matrix, comparison_vector

PRIVATE_HAND_LABELS = [" ".join(hand_names(hand)) for hand in PRIVATE_HANDS]
HAND_INDEX = {label: index for index, label in enumerate(PRIVATE_HAND_LABELS)}

print(f"Loaded {len(PRIVATE_HANDS)} possible private hands.")
```

## Enter a private hand and river board

Tens can be written as either `T` or `10`, for example `Tc` or `10c`.

```python
private_text = "10c Ad"
board_text = "2c 7d 9h Js 3c"

private = make_hand(private_text.split())
board = make_hand(board_text.split())

vector = comparison_vector(private, board)
print("Vector length:", len(vector))
```

## Inspect vector results

```python
opponent_text = "9c Kd"
opponent = make_hand(opponent_text.split())
opponent_label = " ".join(hand_names(opponent))
opponent_index = HAND_INDEX[opponent_label]

print(private_text, "against", opponent_text, "=", vector[opponent_index])
```

```python
# Display every labelled entry in the vector without truncation.
import pandas as pd

result_names = {0: "Loss", 1: "Tie", 2: "Win", "I": "Incompatible"}
vector_table = pd.DataFrame({
    "Vector index": range(len(vector)),
    "Opponent hand": PRIVATE_HAND_LABELS,
    "Value": vector,
    "Result": [result_names[value] for value in vector],
})

with pd.option_context("display.max_rows", None):
    display(vector_table)
```

## Visualise the whole vector

Every one of the 1,326 vector entries is plotted at its original vector index. Nothing is sampled, grouped, wrapped, or omitted. Use the complete table above to match any point to its opponent hand.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch

value_positions = {0: 0, 1: 1, 2: 2, "I": 3}
value_colors = {0: "#d73027", 1: "#fdae61", 2: "#1a9850", "I": "#bdbdbd"}
x = np.arange(len(vector))
y = np.array([value_positions[value] for value in vector])
point_colors = [value_colors[value] for value in vector]

fig, ax = plt.subplots(figsize=(22, 7))
ax.scatter(x, y, c=point_colors, marker="s", s=18, linewidths=0)
ax.set_title(f"{private_text} against every private hand on {board_text}")
ax.set_xlabel("Vector index (all 1,326 opponent hands)")
ax.set_ylabel("Result")
ax.set_xlim(-1, len(vector))
ax.set_ylim(-0.5, 3.5)
ax.set_yticks([0, 1, 2, 3], ["Loss (0)", "Tie (1)", "Win (2)", "Incompatible (I)"])
ax.grid(axis="x", alpha=0.2)
ax.legend(
    handles=[Patch(facecolor=value_colors[value], label=result_names[value]) for value in (0, 1, 2, "I")],
    bbox_to_anchor=(1.01, 1),
    loc="upper left",
)
plt.tight_layout()
plt.show()
```

## Create the full comparison matrix

Rows and columns use `PRIVATE_HAND_LABELS` in the same order.

```python
matrix = comparison_matrix(board)
print("Matrix shape:", len(matrix), "x", len(matrix[0]))
```

```python
# Look up any matrix cell. The result is from the row hand's perspective.
row_hand_text = "10c Ad"
column_hand_text = "9c Kd"

row_label = " ".join(hand_names(make_hand(row_hand_text.split())))
column_label = " ".join(hand_names(make_hand(column_hand_text.split())))
result = matrix[HAND_INDEX[row_label]][HAND_INDEX[column_label]]

print(row_hand_text, "against", column_hand_text, "=", result)
```
