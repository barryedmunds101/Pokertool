# Testing the card bitset module

This notebook checks card conversion, hand creation, and deck behaviour. A successful test cell prints a check mark; a failed assertion stops the cell with an error.

```python
import random

from cards import (
    Deck, FULL_DECK, RANKS, SUITS, card, card_name,
    cards_in, hand_names, make_hand, parse_card,
)

print("PASS: imports successful")
```

## Individual cards

```python
ace_of_spades = parse_card("As")

assert ace_of_spades == card("A", "s")
assert ace_of_spades == 1 << 51
assert card_name(ace_of_spades) == "As"
assert parse_card("2c") == 1

print("Ace of spades integer:", ace_of_spades)
print("Ace of spades bits:   ", f"{ace_of_spades:052b}")
print("PASS: card conversion tests")
```

## Hands as bitsets

```python
hand = make_hand(["As", "Kd", "Th", "2c", "7s"])

assert hand.bit_count() == 5
assert set(hand_names(hand)) == {"As", "Kd", "Th", "2c", "7s"}
assert len(cards_in(hand)) == 5
assert hand & parse_card("As")
assert not hand & parse_card("Ah")

print("Cards:  ", " ".join(hand_names(hand)))
print("Integer:", hand)
print("Bits:   ", f"{hand:052b}")
print("PASS: hand tests")
```

## Full deck and drawing cards

```python
assert len(RANKS) == 13
assert len(SUITS) == 4
assert FULL_DECK.bit_count() == 52

deck = Deck(rng=random.Random(42))
first_hand = deck.draw(5)
second_hand = deck.draw(5)

assert first_hand.bit_count() == 5
assert second_hand.bit_count() == 5
assert first_hand & second_hand == 0  # No card was drawn twice.
assert len(deck) == 42

print("First hand: ", " ".join(hand_names(first_hand)))
print("Second hand:", " ".join(hand_names(second_hand)))
print("Cards remaining:", len(deck))
print("PASS: deck drawing tests")
```

```python
deck.reset(shuffle=False)
assert len(deck) == 52
assert deck.draw(52) == FULL_DECK
assert len(deck) == 0
print("PASS: reset and full-deck tests")
```

## Invalid input checks

```python
def assert_raises(error_type, function, *args, **kwargs):
    try:
        function(*args, **kwargs)
    except error_type:
        return
    raise AssertionError(f"{function.__name__} did not raise {error_type.__name__}")

assert_raises(ValueError, parse_card, "1s")
assert_raises(ValueError, parse_card, "ace of spades")
assert_raises(ValueError, make_hand, ["As", "As"])
assert_raises(ValueError, card_name, 0)
assert_raises(ValueError, Deck().draw, 53)

print("PASS: invalid input tests")
```
